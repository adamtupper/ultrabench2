# Visualize Predicted Masks for Different Models

```python
import csv
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

from ultrabench2 import load_dataset
from ultrabench2.core.rle import rle_to_mask

RESULTS_DIR = Path("/project/ultrabench2/runs/")

plt.rcParams["figure.figsize"] = (6.5, 4)
sns.set_theme(style="whitegrid", context="paper", palette="colorblind", font_scale=0.8)
```

```python
predictions_csvs = {m.name: {} for m in RESULTS_DIR.iterdir() if m.is_dir()}

for model in predictions_csvs:
    tasks = [
        t
        for t in (RESULTS_DIR / model).iterdir()
        if t.is_dir() and t.name.endswith("_segmentation")
    ]

    for task in tasks:
        # Find the predictions.csv file for seed 42 and full fine-tuning
        predictions_csv = next(
            (
                p
                for p in (task / "shot_1.0" / "seed_42").rglob("predictions.csv")
                if "full_finetuning" in str(p)
            ),
            None,
        )
        if predictions_csv:
            predictions_csvs[model][task.name] = predictions_csv
```

```python
dataset = load_dataset(name="AUL")


with open(predictions_csvs["urfm_large"]["aul_mass_segmentation"]) as f:
    reader = csv.DictReader(f)
    urmf_l_predictions = list(reader)

with open(predictions_csvs["dinov3_convnext_large"]["aul_mass_segmentation"]) as f:
    reader = csv.DictReader(f)
    dinov3_convnext_large_predictions = list(reader)

with open(predictions_csvs["echocare"]["aul_mass_segmentation"]) as f:
    reader = csv.DictReader(f)
    echocare_predictions = list(reader)

with open(predictions_csvs["tinyusfm"]["aul_mass_segmentation"]) as f:
    reader = csv.DictReader(f)
    tinyusfm_predictions = list(reader)

fig, axes = plt.subplots(nrows=1, ncols=5)

for i, ax in enumerate(axes.flatten()):
    image = dataset["test.256"][i]["image"]
    mask = dataset["test.256"][i]["mass_mask"]
    urmf_l_predicted_mask = rle_to_mask(
        urmf_l_predictions[i]["encoded_pixels"], height=256, width=256
    )
    dinov3_convnext_large_predicted_mask = rle_to_mask(
        dinov3_convnext_large_predictions[i]["encoded_pixels"], height=256, width=256
    )
    echocare_predicted_mask = rle_to_mask(
        echocare_predictions[i]["encoded_pixels"], height=256, width=256
    )
    tinyusfm_predicted_mask = rle_to_mask(
        tinyusfm_predictions[i]["encoded_pixels"], height=256, width=256
    )

    ax.imshow(image, cmap="gray")
    c_gt = ax.contour(mask, colors="C0", linestyles="dashed", linewidths=0.5)
    c_pred = ax.contour(
        urmf_l_predicted_mask, colors="C1", linestyles="dotted", linewidths=0.5
    )
    dinov3_c_pred = ax.contour(
        dinov3_convnext_large_predicted_mask,
        colors="C2",
        linestyles="solid",
        linewidths=0.5,
    )
    echocare_c_pred = ax.contour(
        echocare_predicted_mask, colors="C3", linestyles="dashdot", linewidths=0.5
    )
    tinyusfm_c_pred = ax.contour(
        tinyusfm_predicted_mask, colors="C4", linestyles="dotted", linewidths=0.5
    )
    h1, _ = c_gt.legend_elements()
    h2, _ = c_pred.legend_elements()
    h3, _ = dinov3_c_pred.legend_elements()
    h4, _ = echocare_c_pred.legend_elements()
    h5, _ = tinyusfm_c_pred.legend_elements()

    if i == len(axes.flatten()) // 2:
        ax.legend(
            [h1[0], h2[0], h3[0], h4[0], h5[0]],
            ["Ground Truth", "URFM-L", "DINOv3", "Echocare", "TinyUSFM"],
            ncols=5,
            loc="lower center",
            bbox_to_anchor=(0.5, -0.2),
        )
    ax.axis("off")

# plt.tight_layout()
plt.savefig("temporary_aul_mass_segmentation.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
from matplotlib.lines import Line2D

MODELS = {
    "urfm_large": "URFM-L",
    "dinov3_convnext_large": "DINOv3",
    "echocare": "Echocare",
    "tinyusfm": "TinyUSFM",
}
TASKS = [
    # (row label, dataset name, task name, mask key)
    ("BUS-BRA", "BUS-BRA", "busbra_segmentation", "tumor_mask"),
    ("TN3K", "TN3K", "tn3k_segmentation", "nodule_mask"),
    ("MMOTU", "MMOTU", "mmotu_segmentation", "tumor_mask"),
    ("AUL Mass", "AUL", "aul_mass_segmentation", "mass_mask"),
]
N_EXAMPLES = 10
CONTOUR_KWARGS = {"linewidths": 0.5, "alpha": 0.5}  # , "linestyles": "dotted"}


def load_predicted_masks(model, task, height=256, width=256):
    """Load the predicted masks of a model for a task, in test set order."""
    with open(predictions_csvs[model][task]) as f:
        return [
            rle_to_mask(row["encoded_pixels"], height=height, width=width)
            for row in csv.DictReader(f)
        ]


def plot_task_row(axes_row, label, dataset_name, task, mask_key):
    """Plot the ground truth and predicted contours for some test examples."""
    test_set = load_dataset(name=dataset_name)["test.256"]
    predicted_masks = {model: load_predicted_masks(model, task) for model in MODELS}

    for i, ax in enumerate(axes_row, start=20):
        example = test_set[i]
        ax.imshow(example["image"], cmap="gray")
        ax.contour(example[mask_key], colors="C0", **CONTOUR_KWARGS)
        for j, model in enumerate(MODELS, start=1):
            ax.contour(predicted_masks[model][i], colors=f"C{j}", **CONTOUR_KWARGS)

        # Hide the axis but keep the y-label for the row title
        ax.set(xticks=[], yticks=[])
        ax.grid(False)
        for spine in ax.spines.values():
            spine.set_visible(False)

    axes_row[0].set_ylabel(label)


# Lay out the panels explicitly (in inches) so the horizontal and vertical gaps
# between the square images are identical
FIG_WIDTH = 6.5
PANEL_GAP = 0.03  # Gap between panels
LABEL_WIDTH = 0.15  # Space on the left for the row labels
LEGEND_HEIGHT = 0.2  # Space at the bottom for the legend

n_rows = len(TASKS)
panel_size = (FIG_WIDTH - LABEL_WIDTH - (N_EXAMPLES - 1) * PANEL_GAP) / N_EXAMPLES
fig_height = n_rows * panel_size + (n_rows - 1) * PANEL_GAP + LEGEND_HEIGHT
grid_left = LABEL_WIDTH / FIG_WIDTH
grid_bottom = LEGEND_HEIGHT / fig_height

fig, axes = plt.subplots(
    nrows=n_rows,
    ncols=N_EXAMPLES,
    figsize=(FIG_WIDTH, fig_height),
    gridspec_kw={
        "left": grid_left,
        "right": 1,
        "bottom": grid_bottom,
        "top": 1,
        # Both are relative to the panel size
        "wspace": PANEL_GAP / panel_size,
        "hspace": PANEL_GAP / panel_size,
    },
)

for axes_row, task_spec in zip(axes, TASKS):
    plot_task_row(axes_row, *task_spec)

legend_handles = [
    Line2D([], [], color=f"C{j}", linewidth=1, alpha=CONTOUR_KWARGS["alpha"])
    for j in range(len(MODELS) + 1)
]
fig.legend(
    legend_handles,
    ["Ground Truth", *MODELS.values()],
    ncols=len(MODELS) + 1,
    loc="upper center",
    bbox_to_anchor=((grid_left + 1) / 2, grid_bottom),  # Centered below the grid
    frameon=False,
)

plt.savefig(
    "predicted_masks_segmentation.png", dpi=300, bbox_inches="tight", pad_inches=0.01
)
plt.savefig(
    "predicted_masks_segmentation.pdf", dpi=300, bbox_inches="tight", pad_inches=0.01
)
plt.show()
```

```python
from matplotlib.patches import Patch
from skimage.color import label2rgb
from matplotlib.colors import ListedColormap

# Test set indices of the examples to show for each task (one column per task)
EXAMPLE_INDICES = {
    "BUS-BRA": [0, 14],
    "TN3K": [1, 19],
    "MMOTU": [10, 18],
    "AUL Mass": [6, 16],
}
CONTOUR_KWARGS = {"linewidths": 0.5, "alpha": 0.5}

color_list = ["#000000", "yellow"]
line_styles = ["dashed", "dotted", "solid", "dashdot"]


def plot_task_column(axes_col, label, dataset_name, task, mask_key, indices):
    """Plot the ground truth and predicted contours for selected examples."""
    test_set = load_dataset(name=dataset_name)["test.256"]
    predicted_masks = {model: load_predicted_masks(model, task) for model in MODELS}

    for idx, ax in zip(indices, axes_col):
        example = test_set[idx]
        ax.imshow(example["image"], cmap="gray")
        # image is (H, W, 3) RGB, mask is (H, W) integer labels
        # print(example[mask_key].shape, example["image"].shape)
        ax.imshow(example[mask_key], cmap=ListedColormap(color_list), alpha=0.3)
        # ax.contour(example[mask_key], colors="C0", **CONTOUR_KWARGS)
        for j, model in enumerate(MODELS, start=0):
            ax.contour(predicted_masks[model][idx], colors=f"C{j}", **CONTOUR_KWARGS)
        ax.axis("off")

    axes_col[0].set_title(label)


# Lay out the panels explicitly (in inches) so the horizontal and vertical gaps
# between the square images are identical
FIG_WIDTH = 6.5
PANEL_GAP = 0.03  # Gap between panels
TITLE_HEIGHT = 0.2  # Space at the top for the column titles
LEGEND_HEIGHT = 0.2  # Space at the bottom for the legend

n_rows = len(next(iter(EXAMPLE_INDICES.values())))
n_cols = len(TASKS)
panel_size = (FIG_WIDTH - (n_cols - 1) * PANEL_GAP) / n_cols
fig_height = (
    n_rows * panel_size + (n_rows - 1) * PANEL_GAP + TITLE_HEIGHT + LEGEND_HEIGHT
)
grid_bottom = LEGEND_HEIGHT / fig_height

fig, axes = plt.subplots(
    nrows=n_rows,
    ncols=n_cols,
    figsize=(FIG_WIDTH, fig_height),
    squeeze=False,
    gridspec_kw={
        "left": 0,
        "right": 1,
        "bottom": grid_bottom,
        "top": 1 - TITLE_HEIGHT / fig_height,
        # Both are relative to the panel size
        "wspace": PANEL_GAP / panel_size,
        "hspace": PANEL_GAP / panel_size,
    },
)

for axes_col, task_spec in zip(axes.T, TASKS):
    plot_task_column(axes_col, *task_spec, EXAMPLE_INDICES[task_spec[0]])

legend_handles = [Patch(facecolor=f"yellow", label="Ground Truth", edgecolor="gray")]
legend_handles += [
    # linestyle=line_styles[j-1], alpha=CONTOUR_KWARGS["alpha"])
    Line2D([], [], color=f"C{j}", linewidth=1)
    for j in range(0, len(MODELS))
]
# legend_handles = [
#     Line2D([], [], color=f"C{j}", linewidth=1, linestyle=line_styles[j-1])#,
# alpha=CONTOUR_KWARGS["alpha"])
#     for j in range(len(MODELS) + 1)
# ]
fig.legend(
    legend_handles,
    ["Ground Truth", *MODELS.values()],
    ncols=len(MODELS) + 1,
    loc="upper center",
    bbox_to_anchor=(0.5, grid_bottom),  # Centered below the grid
    frameon=False,
)

plt.savefig(
    "predicted_masks_segmentation_compact2.png",
    dpi=300,
    bbox_inches="tight",
    pad_inches=0.01,
)
plt.savefig(
    "predicted_masks_segmentation_compact2.pdf",
    dpi=300,
    bbox_inches="tight",
    pad_inches=0.01,
)
plt.show()
```

```python
import numpy as np

OPEN_KIDNEY_PATH = "/project/data/raw/kidneyUS"  # Local dataset, not on the Hub
ANATOMY_TASKS = [
    # (row label, load_dataset kwargs, task name, mask key)
    ("AUL Liver", {"name": "AUL"}, "aul_liver_segmentation", "liver_mask"),
    (
        "CAMUS",
        {"name": "CAMUS", "subset": "segmentation"},
        "camus_segmentation",
        "mask",
    ),
    (
        "Kidney\nRegions",
        {"name": "OpenKidney", "path": OPEN_KIDNEY_PATH},
        "open_kidney_regions_segmentation",
        "regions_mask",
    ),
    (
        "Kidney\nCapsule",
        {"name": "OpenKidney", "path": OPEN_KIDNEY_PATH},
        "open_kidney_capsule_segmentation",
        "capsule_mask",
    ),
    ("PSFHS", {"name": "PSFHS"}, "psfhs_segmentation", "psfh_mask"),
    ("STMUS", {"name": "STMUS"}, "stmus_segmentation", "mask"),
]
# Test set indices of the examples to show for each task (one row per task)
ANATOMY_EXAMPLE_INDICES = {
    label: list(range(N_EXAMPLES)) for label, *_ in ANATOMY_TASKS
}
# Consecutive CAMUS test images are frames of the same sequence, so spread
# the examples across the test set (1,968 images) instead
ANATOMY_EXAMPLE_INDICES["CAMUS"] = list(range(0, 1968, 197))


def load_predicted_label_maps(model, task, height=256, width=256):
    """Load the predicted label maps of a model for a task, keyed by image ID.

    The predictions CSV has one row per image and foreground class, so the
    binary mask of each class is merged into a single map of class IDs.
    """
    label_maps = {}
    with open(predictions_csvs[model][task]) as f:
        for row in csv.DictReader(f):
            label_map = label_maps.setdefault(
                row["image_id"], np.zeros((height, width), dtype=np.uint8)
            )
            mask = rle_to_mask(row["encoded_pixels"], height=height, width=width)
            label_map[mask.astype(bool)] = int(row["class_id"])
    return label_maps


def contour_classes(ax, label_map, color):
    """Draw the outline of every foreground class in a label map."""
    label_map = np.asarray(label_map)
    for class_id in np.unique(label_map[label_map > 0]):
        ax.contour(label_map == class_id, levels=[0.5], colors=color, **CONTOUR_KWARGS)


def plot_anatomy_row(axes_row, label, dataset_kwargs, task, mask_key, indices):
    """Plot the ground truth and predicted contours for selected examples."""
    test_set = load_dataset(**dataset_kwargs)["test.256"]
    predicted = {model: load_predicted_label_maps(model, task) for model in MODELS}

    for idx, ax in zip(indices, axes_row):
        example = test_set[idx]
        image_id = Path(example["filename"]).stem
        ax.imshow(example["image"], cmap="gray")
        contour_classes(ax, example[mask_key], "C0")
        for j, model in enumerate(MODELS, start=1):
            contour_classes(ax, predicted[model][image_id], f"C{j}")

        # Hide the axis but keep the y-label for the row title
        ax.set(xticks=[], yticks=[])
        ax.grid(False)
        for spine in ax.spines.values():
            spine.set_visible(False)

    axes_row[0].set_ylabel(label)


# Lay out the panels explicitly (in inches) so the horizontal and vertical gaps
# between the square images are identical
FIG_WIDTH = 6.5
PANEL_GAP = 0.03  # Gap between panels
LABEL_WIDTH = 0.3  # Space on the left for the (two-line) row labels
LEGEND_HEIGHT = 0.2  # Space at the bottom for the legend

n_rows = len(ANATOMY_TASKS)
panel_size = (FIG_WIDTH - LABEL_WIDTH - (N_EXAMPLES - 1) * PANEL_GAP) / N_EXAMPLES
fig_height = n_rows * panel_size + (n_rows - 1) * PANEL_GAP + LEGEND_HEIGHT
grid_left = LABEL_WIDTH / FIG_WIDTH
grid_bottom = LEGEND_HEIGHT / fig_height

fig, axes = plt.subplots(
    nrows=n_rows,
    ncols=N_EXAMPLES,
    figsize=(FIG_WIDTH, fig_height),
    gridspec_kw={
        "left": grid_left,
        "right": 1,
        "bottom": grid_bottom,
        "top": 1,
        # Both are relative to the panel size
        "wspace": PANEL_GAP / panel_size,
        "hspace": PANEL_GAP / panel_size,
    },
)

for axes_row, task_spec in zip(axes, ANATOMY_TASKS):
    plot_anatomy_row(axes_row, *task_spec, ANATOMY_EXAMPLE_INDICES[task_spec[0]])

legend_handles = [
    Line2D([], [], color=f"C{j}", linewidth=1, alpha=CONTOUR_KWARGS["alpha"])
    for j in range(len(MODELS) + 1)
]
fig.legend(
    legend_handles,
    ["Ground Truth", *MODELS.values()],
    ncols=len(MODELS) + 1,
    loc="upper center",
    bbox_to_anchor=((grid_left + 1) / 2, grid_bottom),  # Centered below the grid
    frameon=False,
)

plt.savefig(
    "predicted_masks_anatomy_segmentation.png",
    dpi=300,
    bbox_inches="tight",
    pad_inches=0.01,
)
plt.savefig(
    "predicted_masks_anatomy_segmentation.pdf",
    dpi=300,
    bbox_inches="tight",
    pad_inches=0.01,
)
plt.show()
```
